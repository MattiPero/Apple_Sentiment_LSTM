# DAILY TWITTER SENTIMENT ANALYSIS WITH VADER

## 1. Data Preparation

In this section, we import the `Pandas` library, load the datasets downloaded from Kaggle, and perform the necessary processing to obtain the Apple tweets dataset required for our analysis.

```python
import pandas as pd
```

```python
df_tweet = pd.read_csv("Tweet.csv")
df_tweet
```

```python
df_company_tweet = pd.read_csv("Company_Tweet.csv")
df_company_tweet
```

```python
df = pd.merge(df_tweet, df_company_tweet, on="tweet_id", how="inner")
df
```

```python
df_apple = df[df["ticker_symbol"] == "AAPL"].copy()
df_apple["post_date"] = pd.to_datetime(df_apple["post_date"], unit="s")
```

```python
df_apple = df_apple.reset_index(drop=True)
```

```python
df_apple
```

```python
df_apple.info()
```

## 2. Sentiment Scoring via VADER

Here we apply the **VADER** algoritm to each Apple tweet to extract the **compound sentiment score**.
The compound score is a continuous metric that ranges from **-1.0** (extreme negative sentiment) to **1.0** (extreme positive sentiment).

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
df_apple["score"] = df_apple["body"].apply(
    lambda x: analyzer.polarity_scores(x)["compound"]
)
```

```python
df_apple
```

## 3. Popularity Feature Engineering & Sentiment Weighting.

In this section, we combine "like_num", "retweet_num" and "comment_num", to create a new feature named "engagement".
Not all tweets carry the same importance; a tweet with high engagement has a significantly stronger impact on market perception than an isolated post.
By multiplying the raw VADER compound score by this new "engagement" metric, we calculate sentiment score that amplifies influential tweets and minimize background noise.

```python
df_apple["engagement"] = (
    df_apple["like_num"] +
    df_apple["retweet_num"] +
    df_apple["comment_num"]
)
```

```python
df_apple["weighted_score"] = df_apple["score"] * df_apple["engagement"]
```

```python
df_sentiment = df_apple.drop(columns=["writer", "tweet_id", "body", "comment_num", "retweet_num", "like_num", "ticker_symbol", "engagement"]).copy()
```

```python
df_sentiment
```

## 4. Daily Data Aggregation

Since financial stock data is recorded on a daily basis, we aggregate our tweet-level metrics.
we group the data by date to compute the average daily weighted sentiment and the total daily tweet volume.

```python
df_sentiment["day"] = df_sentiment["post_date"].dt.date

df_sentiment["daily_tweet_count"] = (
    df_sentiment.groupby("day").transform("size")
)
```

```python
df_sentiment["daily_score"] = (
    df_sentiment.groupby("day")["weighted_score"].transform("mean")
)
```

```python
df_sentiment
```

```python
df_daily_sentiment = df_sentiment[
    ["day", "daily_tweet_count", "daily_score"]
].drop_duplicates().reset_index(drop=True)
df_daily_sentiment
```
